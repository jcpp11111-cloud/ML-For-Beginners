# **UFO SIGHTINGS**
##### 80,000 UFO SIGHTINGS GATHERED BY NUFORC

```python
import pandas as pd
import numpy as np
import matplotlib as plt
import pickle as pkl
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.linear_model import LogisticRegression

ufos_Raw = pd.read_csv('./data/ufos.csv')
```

```python
ufos = pd.DataFrame({'Seconds' : ufos_Raw['duration (seconds)'], 'Country' : ufos_Raw['country'], 'Latitude' : ufos_Raw['latitude'], 'Longitude' : ufos_Raw['longitude']})
ufos.Country.value_counts()
```

Now we are gonna delete all nan values or incomplete values for us to be easier to read and manipulate the data

```python
ufos.dropna(inplace=True)
ufos = ufos[(ufos['Seconds'] >=1) & (ufos['Seconds'] <= 60)]
ufos.info()
```

```python
ufos['Country'] =LabelEncoder().fit_transform(ufos['Country'])
ufos.head()
```

```python
sf = ['Latitude', 'Seconds', 'Longitude']
X = ufos[sf]
y = ufos['Country']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=0)
model = LogisticRegression(max_iter=100000)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

#### **Lets see how accurate our model is:**

```python
print(classification_report(y_test, predictions))
print('Prediction Labels: ', predictions)
print('Acurracy: ', accuracy_score(y_test, predictions))
```

```python
model_filename = 'ufo-model.pkl'
with open('ufo-model.pkl','wb') as f:
    pkl.dump(model, f)
print(model.predict([[50,44,-12]]))
```
